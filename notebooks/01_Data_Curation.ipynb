# Data Curation


## Import Libraries


```python
import numpy as np
import pandas as pd
```

## Load Dataset


```python
df=pd.read_csv("../data/input/IC50LibA.csv")
df
```

## Dataset Overview


```python
df.columns
```

```python
df.info()
```

## Check Duplicate Ligands


```python
duplicate=df[df.duplicated(subset='Smile',keep= False)]
duplicate
```

## Remove Duplicate Ligands

Duplicate molecules were identified based on SMILES strings
When duplicate SMILES existed, the median IC50 value was retained.


```python
df= df.groupby("Smile", as_index=False).agg({'ID':'first', 'IC50(nM)': 'median'})
df
```

```python
print(f"Initial molecules : 180")
print(f"After curation    : {len(df)}")
```

## Calculate pIC50
The IC50 values (nM) were converted to pIC50 using

pIC50 = -log10(IC50 × 10⁻⁹)

```python
df['pIC50']=-np.log10(df['IC50(nM)']*1e-9)
df
```
